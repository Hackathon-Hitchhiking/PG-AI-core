# Local HF Image Agent

```python
from agents.image import ImageAgent, ImageModelConfig, GenerationRequest
config = ImageModelConfig(
    model_type="diffusers",
    model_name="OFA-Sys/small-stable-diffusion-v0",
    device="mps",
    pipeline_kwargs={
        "attention_slicing": True,
        "requires_safety_checker": False
    }
)

agent = ImageAgent(config)
```

```python
request = GenerationRequest(
    prompt="Programmer on toilet use docker",
    width=128,
    height=128,
    num_inference_steps=50,
    guidance_scale=7.5
)

image = agent.generate(request)

image.show()
image.save("output_hf_local.png")
```

# API Image Agent

```python
from agents.image import ImageAgent, ImageModelConfig, GenerationRequest
config = ImageModelConfig(
    model_type="api",
    api_base="https://image.pollinations.ai/prompt/",
    model_kwargs={"method": "GET"}
)

agent = ImageAgent(config)
```

```python
request = GenerationRequest(
    prompt="senpai otaku",
    negative_prompt="Bad quality, low quality",
    width=512,
    height=512,
    num_inference_steps=2,
    guidance_scale=7.5
)

image = agent.generate(request)
image.save("pollinations_test_API.png")
image.show()
```

# OpenAI API Image Agent

```python
from agents.image import ImageAgent, ImageModelConfig, GenerationRequest
import os
os.environ["OPENAI_API_KEY"] = "<YOUR-OPENAI-API-KEY>"

config = ImageModelConfig(
    model_type="openai",
    api_key="<YOUR-OPENAI-API-KEY>"
)

agent = ImageAgent(config)

request = GenerationRequest(
    prompt="A sleek futuristic cityscape underwater",
    width=512,     # DALL-E can generate 256x256, 512x512, or 1024x1024
    height=512,    # so pick a supported size
)

image = agent.generate(request)
image.show()
```
