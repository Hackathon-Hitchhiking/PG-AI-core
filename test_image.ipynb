# Local HF Image Agent

```python
from schemas.image import DiffusersConfig, GenerationRequest
from agents.image.agent import ImageAgent

# Для Diffusers
config = DiffusersConfig(
    model_name="OFA-Sys/small-stable-diffusion-v0",
    device="mps",
    torch_dtype="float16",
)

# Инициализация агента
agent = ImageAgent(config)

# Запрос на генерацию
request = GenerationRequest(
    prompt="A cute cartoon fox sitting in a meadow, digital art",
    width=768,
    height=768,
    num_inference_steps=25,
    guidance_scale=9.0,
    seed=42
)

# Генерация изображения
try:
    image = agent.generate(request)
    image.save("diffusers_output.jpg")
    print("Image generated successfully with Diffusers!")
except Exception as e:
    print(f"Generation failed: {e}")
```
